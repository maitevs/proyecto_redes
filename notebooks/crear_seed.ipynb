## Dieta intestinal

Además de mtym_e necesitamos todos los metabolitos que son parte de el intestino. Buscamos encontrar el medio mínimo derivado de la intersección de metabolitos de intercambio de los modelos de la comunidad

```python
import cobra, glob, os
from collections import Counter

c = Counter()
for f in sorted(glob.glob('sbml/*.xml')):
    try:
        m = cobra.io.read_sbml_model(f)
    except Exception:
        continue
    for r in m.reactions:
        if r.id.startswith('EX_'):
            for met in r.metabolites:
                c[met.id] += 1

n = len([f for f in glob.glob('sbml/*.xml')])
print(f"{n} modelos\n")
for mid, k in c.most_common():
    if k >= 8:          # presente en casi todos
        print(f"{k:3d}  {mid}")
```

```python
EXCLUIR = {
    'biomass_c', 'o2_e',                      # no nutriente / anaerobio
    'succ_e', 'ac_e', 'for_e', 'lac_L_e',     # productos de fermentación
    'lac_D_e', 'fum_e',
    'cd2_e', 'hg2_e', 'pb2_e',                # metales tóxicos
    'digoxin_e', 'chlphncl_e', 'nchlphncl_e', # otros fármacos
    'czp_e', '7a_czp_e', 'anzp_e', 'nzp_e',
    'r406_e', 'dfdcytd_e', 'dfduri_e',
    'dpmn_e', 'mtym_e',                       # la cascada: NO van
}

seeds = [mid for mid, k in c.items() if k >= 8 and mid not in EXCLUIR]
seeds.append('levdopa_e')                      # el fármaco de interés

with open('m2m/seeds.txt', 'w') as f:
    for s in sorted(set(seeds)):
        f.write(f"M_{s}\n")

with open('m2m/targets.txt', 'w') as f:
    f.write("M_mtym_e\n")

print(f"{len(set(seeds))} seeds escritos")
```
