```python
path = "/Users/maitevillagran/Desktop/proyecto_redes/mis15.txt"
# Source - https://stackoverflow.com/a/70213858

f = open(path)
content = f.read()
f.close()


for line in content.split('\n'):
    for char in line:
        if char in line:
            if char in "_":
                line = line.replace(char, " ")
    print(line)
    
```
